# Operational Source Overview

Notebook này khảo sát các operational source entities hiện có để xác định:

- schema
- primary key
- foreign key
- mutable fields
- timestamp fields
- dependency giữa các entity
- mức độ sẵn sàng của Bronze

Các CSV trong Git Folder chỉ được sử dụng làm source reference cho quá trình khảo sát.

Operational Silver production pipeline không đọc trực tiếp từ các CSV này.

```python
import os

account_name = os.getenv("AZURE_STORAGE_ACCOUNT_NAME")
tenant_id = os.getenv("AZURE_TENANT_ID")
client_id = os.getenv("AZURE_CLIENT_ID")
client_secret = os.getenv("AZURE_CLIENT_SECRET")

endpoint = f"{account_name}.dfs.core.windows.net"

spark.conf.set(
    f"fs.azure.account.auth.type.{endpoint}",
    "OAuth"
)

spark.conf.set(
    f"fs.azure.account.oauth.provider.type.{endpoint}",
    "org.apache.hadoop.fs.azurebfs.oauth2.ClientCredsTokenProvider"
)

spark.conf.set(
    f"fs.azure.account.oauth2.client.id.{endpoint}",
    client_id
)

spark.conf.set(
    f"fs.azure.account.oauth2.client.secret.{endpoint}",
    client_secret
)

spark.conf.set(
    f"fs.azure.account.oauth2.client.endpoint.{endpoint}",
    f"https://login.microsoftonline.com/{tenant_id}/oauth2/token"
)
```

```python
project_root = os.getcwd()
print(project_root)
```

```python
from pathlib import Path

PROJECT_ROOT = Path(os.getcwd()).parents[2]

print(PROJECT_ROOT)
```

```python
OLIST_RAW_DIR = (
    PROJECT_ROOT
    / "data"
    / "raw"
    / "olist"
)

print(OLIST_RAW_DIR)
```

```python
SOURCE_FILES = {
    "customers": "olist_customers_dataset.csv",
    "orders": "olist_orders_dataset.csv",
    "order_items": "olist_order_items_dataset.csv",
    "payments": "olist_order_payments_dataset.csv",
    "reviews": "olist_order_reviews_dataset.csv",
    "products": "olist_products_dataset.csv",
    "sellers": "olist_sellers_dataset.csv",
    "geolocation": "olist_geolocation_dataset.csv",
}
```

```python
def load_source_csv(file_name: str):
    path = OLIST_RAW_DIR / file_name

    if not path.exists():
        raise FileNotFoundError(
            f"Source CSV not found: {path}"
        )

    return (
        spark.read
        .format("csv")
        .option("header", True)
        .option("inferSchema", True)
        .load(f"file:{path}")
    )
```

```python
df_orders_source = load_source_csv(
    SOURCE_FILES["orders"]
)
df_customers_source = load_source_csv(
    SOURCE_FILES["customers"]
)
df_order_items_source = load_source_csv(
    SOURCE_FILES["order_items"]
)
df_payments_source = load_source_csv(
    SOURCE_FILES["payments"]
)
df_reviews_source = load_source_csv(
    SOURCE_FILES["reviews"]
)
df_products_source = load_source_csv(
    SOURCE_FILES["products"]
)
df_sellers_source = load_source_csv(
    SOURCE_FILES["sellers"]
)
df_geolocation_source = load_source_csv(
    SOURCE_FILES["geolocation"]
)
```
